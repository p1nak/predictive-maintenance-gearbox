```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
healthy_path = "Healthy Data"
broken_path = "BrokenTooth Data"
```

```python
healthy_files = sorted(os.listdir(healthy_path))
broken_files = sorted(os.listdir(broken_path))
```

```python
print("Healthy Files :", len(healthy_files))
print("Broken Files :", len(broken_files))
```

```python
print("First 5 Healthy Files")
print(healthy_files[:5])

print()

print("First 5 Broken Files")
print(broken_files[:5])
```

```python
healthy_sample = np.loadtxt(os.path.join(healthy_path, healthy_files[0]))
```

```python
np.isnan(healthy_sample).sum()
```

```python
healthy_sample
```

```python
print(type(healthy_sample))
```

```python
print(healthy_sample.shape)
```

```python
print(healthy_sample.dtype)
```

```python
healthy_sample[:10]
```

```python
plt.figure(figsize=(14,5))

plt.plot(healthy_sample)

plt.title("Healthy Gearbox Vibration Signal")

plt.xlabel("Sample Number")

plt.ylabel("Amplitude")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(healthy_sample, bins=50)

plt.title("Amplitude Distribution")
plt.xlabel("Amplitude")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(5,6))

plt.boxplot(healthy_sample)

plt.title("Healthy Signal Box Plot")
plt.ylabel("Amplitude")

plt.show()
```

```python
broken_sample = np.loadtxt(os.path.join(broken_path, broken_files[0]))
```

```python
np.isnan(broken_sample).sum()
```

```python
broken_sample
```

```python
plt.figure(figsize=(14,5))

plt.plot(broken_sample)

plt.title("Broken Tooth Gearbox Signal")

plt.xlabel("Sample Number")

plt.ylabel("Amplitude")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(broken_sample, bins=50)

plt.title("Broken Signal Amplitude Distribution")
plt.xlabel("Amplitude")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(5,6))

plt.boxplot(broken_sample)

plt.title("Broken Signal Box Plot")
plt.ylabel("Amplitude")

plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.plot(healthy_sample, label="Healthy")

plt.plot(broken_sample,label="Broken",alpha=0.7)

plt.legend()

plt.xlabel("Sample Number")

plt.ylabel("Amplitude")

plt.title("Healthy vs Broken Gearbox Signal")

plt.show()
```

```python
healthy_lengths = []

for file in healthy_files:

    signal = np.loadtxt(os.path.join(healthy_path, file))

    healthy_lengths.append(len(signal))

pd.Series(healthy_lengths).describe()
```

```python
summary = pd.DataFrame({
    "Class": ["Healthy", "Broken"],
    "Number of Files": [len(healthy_files), len(broken_files)],
    "Signal Length": [len(healthy_sample), len(broken_sample)]
})

summary
```
